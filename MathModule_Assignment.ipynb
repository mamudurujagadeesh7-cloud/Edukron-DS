```python
#30
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("\n1. Arithmetic\n2. Comparison\n3. Logical\n4. Membership\n5. Identity\n6. Bitwise")

choice = int(input("Choose operation: "))

if choice == 1:
    print("Addition:", a + b)
    print("Subtraction:", a - b)
    print("Multiplication:", a * b)
    print("Division:", a / b)

elif choice == 2:
    print("a > b:", a > b)
    print("a < b:", a < b)
    print("a == b:", a == b)

elif choice == 3:
    print("Both positive:", a > 0 and b > 0)
    print("Any positive:", a > 0 or b > 0)
    print("Not equal:", not(a == b))

elif choice == 4:
    numbers = [a, b, 10, 20]
    print("a in list:", a in numbers)
    print("b not in list:", b not in numbers)

elif choice == 5:
    x = a
    y = a
    print("x is y:", x is y)
    print("x is not b:", x is not b)

elif choice == 6:
    print("AND:", a & b)
    print("OR:", a | b)
    print("XOR:", a ^ b)

else:
    print("Invalid choice")
```
