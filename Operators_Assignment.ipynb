```python
#1
print("1.Addition")
print("2.Subtraction")
print("3.Multiplication")
print("4.Division")
print("5.Floor division")
print("6.Modulus")
print("7.Exponentiation")

choice = float(input("Enter the chocice : "))
x = float(input("Enter the  Number :"))
y = float(input("Enter the  Number : "))

if choice == 1:
  Add = x + y
  print(f"Addition of {x} , {y} is", Add)

elif choice ==2:
  Sub = x - y
  print(f"subtraction of {x} , {y} is", Sub)

elif choice == 3:
  Mul = x * y
  print(f"Multiplication of {x} , {y} is", Mul)

elif choice == 4:
  Div = x / y
  print(f"Division of {x} , {y} is", Div)

elif choice == 5:
  Floor_Div = x // y
  print(f"floor_division of {x} , {y} is", Floor_Div)

elif choice == 6:
  Mod = x % y
  print(f"Modulus of {x} , {y} is", Mod)

elif choice == 7:
  Exp = x ** y
  print(f"exponentiation of {x} , {y} is", Exp)

else:
  print("Invalid choice")
```

```python
#30
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("\n1. Arithmetic\n2. Comparison\n3. Logical\n4. Membership\n5. Identity\n6. Bitwise")

choice = int(input("Choose operation: "))

if choice == 1:
    print("Addition:", a + b)
    print("Subtraction:", a - b)
    print("Multiplication:", a * b)
    print("Division:", a / b)

elif choice == 2:
    print("a > b:", a > b)
    print("a < b:", a < b)
    print("a == b:", a == b)

elif choice == 3:
    print("Both positive:", a > 0 and b > 0)
    print("Any positive:", a > 0 or b > 0)
    print("Not equal:", not(a == b))

elif choice == 4:
    numbers = [a, b, 10, 20]
    print("a in list:", a in numbers)
    print("b not in list:", b not in numbers)

elif choice == 5:
    x = a
    y = a
    print("x is y:", x is y)
    print("x is not b:", x is not b)

elif choice == 6:
    print("AND:", a & b)
    print("OR:", a | b)
    print("XOR:", a ^ b)

else:
    print("Invalid choice")
```
