```python
# Part 10: Nested Report Generator
def create_department_report(dept):
    department_staff = list(filter(
        lambda emp: emp["department"].lower() == dept.lower(),employees))
    def count_staff():
        return len(department_staff)
    def salary_total():
        return sum(emp["salary"] for emp in department_staff)
    def salary_average():
        if count_staff() == 0:
            return 0
        return salary_total() / count_staff()
    def performance_average():
        if count_staff() == 0:
            return 0
        return sum(emp["performance_score"]
            for emp in department_staff) / count_staff()
    return {
        "Employee Count": count_staff(),
        "Total Salary": salary_total(),
        "Average Salary": salary_average(),
        "Average Performance": performance_average()
    }
dept_input = input("Enter Department Name: ")
report_data = create_department_report(dept_input)
for key, value in report_data.items():
    print(f"{key}: {value}")
```
