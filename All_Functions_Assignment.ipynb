```python
#19
def find_list_maximum(numbers):
    if len(numbers) == 0:
        return "list is empty."
    maximum = numbers[0]
    for num in numbers:
        if num > maximum:
            maximum = num
    return maximum
numbers = list(map(int,input("Enter the numbers :").split(" ")))
result = find_list_maximum(numbers)
print("Maximum:", result)
```
